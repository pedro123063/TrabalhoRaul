```python
import os
import pandas as pd
import matplotlib.pyplot as plt


def list_and_plot_tickers(csv_file="stock_prices_daily.csv", output_path="output/lista_tickers.png"):
    """
    Lê a coluna 'Ticker' do CSV, remove duplicatas com set, 
    ordena e gera uma tabela visual elegante em PNG.
    """
    print(f"Lendo '{csv_file}'...")
    
    try:
        # Lemos apenas a coluna 'Ticker' para o processo ser extremamente rápido
        df = pd.read_csv(csv_file, usecols=["Ticker"])
        
        # Usamos o 'set' para garantir a unicidade de forma rápida e simples
        tickers_unicos = sorted(list(set(df["Ticker"].dropna())))
        
    except FileNotFoundError:
        print(f"\n[Erro] O arquivo '{csv_file}' não foi encontrado na raiz do projeto.")
        return
    except ValueError:
        # Caso a coluna 'Ticker' não exista com esse nome exato
        df_cabecalho = pd.read_csv(csv_file, nrows=1)
        print(f"\n[Erro] Coluna 'Ticker' não encontrada.")
        print(f"Colunas disponíveis no seu CSV: {df_cabecalho.columns.tolist()}")
        return

    total_tickers = len(tickers_unicos)
    print(f"Sucesso! Encontrados {total_tickers} tickers únicos: {tickers_unicos}")

    # Criamos um DataFrame simples para alimentar o gerador de tabelas
    df_tickers = pd.DataFrame(tickers_unicos, columns=["Tickers no CSV"])

    # Ajustamos dinamicamente o tamanho da imagem de acordo com a quantidade de tickers
    altura_fig = max(2, total_tickers * 0.4 + 1)
    
    # Criamos a figura do matplotlib sem eixos
    fig, ax = plt.subplots(figsize=(4, altura_fig))
    ax.axis('off')

    # Desenhamos a tabela
    tabela = ax.table(
        cellText=df_tickers.values,
        colLabels=df_tickers.columns,
        cellLoc='center',
        loc='center'
    )

    # Estilização visual (igual à que fizemos no pipeline de gráficos)
    tabela.auto_set_font_size(False)
    tabela.set_fontsize(12)
    tabela.scale(1.2, 1.8)  # Altura das células

    for (row, col), cell in tabela.get_celld().items():
        if row == 0:
            cell.set_text_props(weight='bold', color='white')
            cell.set_facecolor('#1e3d59')  # Azul escuro elegante
        else:
            cell.set_facecolor('#f5f5f5' if row % 2 == 0 else 'white')
        cell.set_edgecolor('#cccccc')

    # Salva a tabela como imagem
    os.makedirs(os.path.dirname(output_path), exist_ok=True)
    plt.savefig(output_path, bbox_inches='tight', dpi=300)
    plt.close()
    
    print(f"\n[OK] Tabela gráfica salva em: '{output_path}'")


if __name__ == "__main__":
    list_and_plot_tickers()
```
